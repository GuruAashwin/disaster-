## Findings -> Cleaning Rules
1. **Names**: We observe initials like 'Arun.K' and name order swaps. **Rule**: Lowercase, strip punctuation, sort tokens alphabetically for a baseline metric.
2. **Ages**: High variance in 'found' records (approximations). **Rule**: Convert all ages to `age_min` and `age_max` ranges. Missing age shouldn't penalize.
3. **Transliteration**: Tamil scripts exist. **Rule**: Pass text through indic-transliteration to map to English alphabet before fuzzy matching.
4. **Class Balance**: Hard negatives (same name, different location) require location distance metrics (Haversine) to filter properly.
